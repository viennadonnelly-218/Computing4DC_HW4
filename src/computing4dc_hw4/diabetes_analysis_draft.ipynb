```python
from diabetes_library.importing.loading import load_data, split_data
from diabetes_library.preprocessing.cleaning import (
    drop_missing_rows,
    fill_missing_with_mean,
)
from diabetes_library.preprocessing.encoding import add_one_hot_dummies, encode_binary
from diabetes_library.modeling.training import train_model, add_predictions
from diabetes_library.modeling.evaluation import compute_roc_auc

import os
```

```python
DATA_PATH =  os.getcwd() + "/data/sample_diabetes_mellitus_data.csv"
FEATURES = [
    "age",
    "height",
    "weight",
    "aids",
    "cirrhosis",
    "hepatic_failure",
    "immunosuppression",
    "leukemia",
    "lymphoma",
    "solid_tumor_with_metastasis",
]
TARGET = "diabetes_mellitus"
```

```python
df = load_data(DATA_PATH)
train, test = split_data(df, test_size=0.2)
```

```python
def preprocess(df):
    """Apply steps c to f to a DataFrame."""
    df = drop_missing_rows(df, ["age", "gender", "ethnicity"])  # c
    df = fill_missing_with_mean(df, ["height", "weight"])  # d
    df = add_one_hot_dummies(df, "ethnicity")  # e
    df = encode_binary(df, "gender", "M", "gender_male")  # f
    return df


train = preprocess(train)
test = preprocess(test)
```

```python
model = train_model(train, FEATURES, TARGET)  # g

train = add_predictions(train, model, FEATURES)  # h
test = add_predictions(test, model, FEATURES)

print(f"Train ROC AUC: {compute_roc_auc(train, TARGET):.3f}")  # i
print(f"Test ROC AUC:  {compute_roc_auc(test, TARGET):.3f}")
```
